```python
from pathlib import Path
import pandas as pd

# Directorio actual
directorio = Path(".")

# Buscar archivos parquet
archivos_parquet = list(directorio.glob("*.parquet"))

# Verificar si existen archivos
if not archivos_parquet:
    print("No se encontraron archivos .parquet en el directorio actual.")
else:
    print(f"Se encontraron {len(archivos_parquet)} archivos parquet.\n")

    # Iterar sobre cada archivo
    for archivo in archivos_parquet:
        print("=" * 80)
        print(f"Archivo: {archivo.name}")
        print("-" * 80)

        try:
            # Leer solo metadata / estructura
            df = pd.read_parquet(archivo)

            # Mostrar columnas
            print(f"Cantidad de columnas: {len(df.columns)}\n")

            for i, col in enumerate(df.columns, start=1):
                print(f"{i:03d}. {col}")

            # Opcional: mostrar tipos de datos
            print("\nTipos de datos:")
            print(df.dtypes)

        except Exception as e:
            print(f"Error leyendo {archivo.name}: {e}")

        print("\n")
```
