```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 200)
pd.set_option("display.max_colwidth", 200)
```

```python
ROOT = Path("../../").resolve()
DATA_DIR = ROOT / "data" / "normalized"

CODEQL_PATH = DATA_DIR / "code_vulnerabilities.parquet"
DEPENDENCIES_PATH = DATA_DIR / "dependencies.parquet"
DEPENDENCY_VULNS_PATH = DATA_DIR / "dependency_vulnerabilities.parquet"
REPOSITORIES_PATH = DATA_DIR / "repositories.parquet"

print(DATA_DIR.resolve())
```

```python
codeql_df = pd.read_parquet(CODEQL_PATH)
dependencies_df = pd.read_parquet(DEPENDENCIES_PATH)
dependency_vulns_df = pd.read_parquet(DEPENDENCY_VULNS_PATH)
repositories_df = pd.read_parquet(REPOSITORIES_PATH)
```

```python
datasets = {
    "repositories": repositories_df,
    "codeql": codeql_df,
    "dependencies": dependencies_df,
    "dependency_vulnerabilities": dependency_vulns_df,
}

for name, df in datasets.items():
    print("=" * 80)
    print(f"DATASET: {name}")
    print("=" * 80)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nDtypes:")
    print(df.dtypes)

    print("\nPreview:")
    display(df.head(3))
```

```python
print("=" * 80)
print("BASIC COUNTS")
print("=" * 80)

print(f"Repositories: {len(repositories_df):,}")

if "repository" in codeql_df.columns:
    print(
        f"Repositories with CodeQL findings: {codeql_df['repository'].nunique():,}")

print(f"CodeQL findings: {len(codeql_df):,}")

if "package" in dependencies_df.columns:
    print(f"Unique packages: {dependencies_df['package'].nunique():,}")

if "vulnerability_id" in dependency_vulns_df.columns:
    print(
        f"Unique CVEs: {dependency_vulns_df['vulnerability_id'].nunique():,}")

print(f"Dependency vulnerabilities: {len(dependency_vulns_df):,}")
```

```python
def analyze_nulls(df, name):
    print("=" * 80)
    print(f"NULL ANALYSIS: {name}")
    print("=" * 80)

    nulls = df.isnull().sum()
    null_percentage = (nulls / len(df)) * 100

    result = pd.DataFrame({
        "null_count": nulls,
        "null_percentage": null_percentage
    })

    result = result.sort_values(
        by="null_percentage",
        ascending=False
    )

    display(result)


for name, df in datasets.items():
    analyze_nulls(df, name)
```

```python
critical_fields = {
    "codeql": [
        "repository",
        "severity",
        "rule_id",
        "file",
    ],
    "dependencies": [
        "repository",
        "package",
        "version",
    ],
    "dependency_vulnerabilities": [
        "repository",
        "package",
        "vulnerability_id",
        "severity",
    ],
}

for dataset_name, fields in critical_fields.items():
    df = datasets[dataset_name]

    print("=" * 80)
    print(f"CRITICAL FIELD VALIDATION: {dataset_name}")
    print("=" * 80)

    for field in fields:
        if field not in df.columns:
            print(f"[MISSING COLUMN] {field}")
            continue

        missing = df[field].isnull().sum()

        print(
            f"{field:<25} -> "
            f"missing: {missing:,} "
            f"({missing / len(df) * 100:.2f}%)"
        )
```
